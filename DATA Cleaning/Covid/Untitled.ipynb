```python
# Death VS conformed
sns.scatterplot(x="Deaths",y="Confirmed",hue="Recovered",data=df)
plt.title("Confirmed vs Deaths")
plt.show()
```

```python
cases_under_who=df.groupby("WHO Region")["Confirmed"].sum().sort_values(ascending=False)
```

```python
sns.barplot(x=cases_under_who.index,y=cases_under_who.values,)
plt.ylabel("Total Cases")
plt.xlabel("Who Regions")
plt.title("Total Cases Under Who Region")
plt.xticks(rotation=45)
plt.show()
```

```python
# Active vs Recovered Cases
df["Active"].hist(bins=30)
plt.title("Active Cases Distribution")
plt.show()
```

```python
# Weekly Growth Analysis By Country
growth=df.sort_values("1 week % increase",ascending=False).head(10)
plt.barh(growth["Country/Region"],growth["1 week % increase"])
```
