```python
import kagglehub
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns 
%matplotlib inline
```

```python
# Download latest version

path= kagglehub.dataset_download(r"iamsouravbanerjee/customer-shopping-trends-dataset")
data=os.listdir(path)
csv_file=os.path.join(path,data[0])
data=pd.read_csv(csv_file)
data.head()
```

```python
data['Age'].plot(kind='hist')
plt.title('Histogram of age')
plt.show()
```

```python
data['Gender'].value_counts().plot(kind='bar')
plt.title('bar plot of gender distribution')
```

```python
df=data
df.columns
```

```python
df.groupby('Category')['Purchase Amount (USD)'].mean().sort_values(ascending=False).head(2)
```

```python
df.sort_values('Review Rating',ascending=False,inplace=True)
```

```python
df['Subscription Status'].value_counts().plot(kind='pie')
```

```python
df.head(75).plot(x='Previous Purchases',y='Review Rating',kind='scatter')
plt.title('Scatter plot previous purchases and review rating')
plt.show()
```
