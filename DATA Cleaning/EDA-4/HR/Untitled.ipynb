```python
# Attrition by JobRole
attr_by_role=df.groupby("JobRole")["Attrition"].size()
```

```python
attr_by_role.values
```

```python
attrition_role = pd.crosstab(df['JobRole'], df['Attrition'], normalize='index')
attrition_role.plot(kind='bar', stacked=True)
plt.xticks(rotation=45)
plt.show()
```
