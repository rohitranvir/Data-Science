# fill with random value*

```python
d2={
    'Names':['Ramesh','Suresh',np.nan,'Mahesh'],
    'Age':[20,22,24,np.nan],
    'City':[np.nan,'Blr','Pune','Mumbai']
}
df2=pd.DataFrame(d2)
df2.fillna(0)
```

```python
d2={
    'Names':['Ramesh','Suresh',np.nan,'Mahesh'],
    'Age':[20,22,24,np.nan],
    'City':[np.nan,'Blr','Pune','Mumbai']
}
df2=pd.DataFrame(d2)
df2['Names'].fillna(df2['Names'].mode())
df2['Age'].fillna(df2['Age'].mean().astype(int))
```
