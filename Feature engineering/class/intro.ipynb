# Encoding
- Encoding means convert categorical data into numerical data
- ML/DL or any model only works based on maths
- Model can not understand English characters
- So it is very important to convert categorical label tonumerical values
- for example gender column has two labels
    - Male
    - Female
- Then Female represent as 0 , Male represent  as 1

```python
df["case_status"].unique()
```

```python
df["case_status"]=df["case_status"].map({"Certified":0,"Denied":1})
```

```python
num_col=df.select_dtypes(include=("int","float"))
```

```python
uni=df["continent"].unique()
```

```python
uni
```

```python
dic={uni[i] : i for i in range(len(uni))}    
```

```python
df["continent"]=df["continent"].map(dic)
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
```

```python
df["case_status"]=le.fit_transform(df["case_status"])
```

```python
for i in df[1:]:
    df[i]=le.fit_transform(df[i])
```
