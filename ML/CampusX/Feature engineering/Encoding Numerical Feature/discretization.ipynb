# Discretization (also called binning) means converting continuous numerical values into discrete categories (bins).

# 1️⃣ What is Discretization?
It transforms a continuous feature like:
- Age: 18, 23, 45, 62

- TO

- Age Group: Young, Adult, Senior

- Or

- 0, 1, 2





# Why use Discretization:

1. To handle Outliers
2. To improve the value spread
