```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
```

```python
ct=ColumnTransformer(
    transformers=(
        ('gender',OneHotEncoder(drop='first'),['gender']),
        ('review',OneHotEncoder(drop='first'),['review']),
        ('education',OneHotEncoder(drop='first'),['education'])
    )
    ,remainder='passthrough'
)
x=df.drop('purchased',axis=1)
X_encoded=ct.fit_transform(df.drop(['purchased'],axis=1))
```

```python
feature_names = ct.get_feature_names_out()

X_encoded_df = pd.DataFrame(
    X_encoded,
    columns=feature_names,
    index=df.index
)
```

```python
X_encoded_df = X_encoded_df.astype(int)
```
