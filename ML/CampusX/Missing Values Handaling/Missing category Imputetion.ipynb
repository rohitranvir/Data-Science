```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("train.csv")
```

```python
df.columns
```

```python
df=df[["FireplaceQu","GarageQual","SalePrice"]]
```

```python
df.head()
```

```python
df.isnull().mean()*100
```

```python
df["GarageQual"].value_counts()
```

```python
df["GarageQual"].value_counts().sort_values(ascending=False).plot(kind="bar")
plt.xlabel("GarageQual")
plt.ylabel("Number of Houses")
plt.show()
```

```python
df["GarageQual"].mode()
```

```python
df[df["GarageQual"]=='TA']["SalePrice"].plot(kind="kde")
df[df["GarageQual"].isnull()]["SalePrice"].plot(kind="kde")
labels=["House With TA","House With NA"]
plt.legend(labels)
```

```python
temp=df[df["GarageQual"]=="TA"]["SalePrice"]
```

```python
df["GarageQual"].fillna("TA",inplace=True)
```

```python
df["GarageQual"].value_counts().plot(kind='bar')
```

```python
ax=plt.subplot(111)
temp.plot(kind="kde",ax=ax)
df[df["GarageQual"]=='TA']["SalePrice"].plot(kind="kde",ax=ax)
labels=["origional variable",'Impute variable']
plt.legend(labels)
```
