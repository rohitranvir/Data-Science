```python
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn. impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn. preprocessing import MinMaxScaler
from sklearn. tree import DecisionTreeClassifier
```

```python
df=sns.load_dataset('titanic')
```

```python
df.drop(columns=['class','who','adult_male','deck','embark_town','alive','alone'],inplace=True)
```

```python
# Step 1
X_train,X_test,y_train,y_test=train_test_split(df.drop(columns=['survived']),df['survived'],test_size=0.2,random_state=42)
```

```python
X_train.head(2)
```

```python
y_train.head()
```

```python
df.isnull().sum()
```

```python
# Applying Stratigic
si_age=SimpleImputer(strategy='mean')
si_embarked=SimpleImputer(strategy='most_frequent')
X_train_age=si_age.fit_transform(X_train[['age']])
X_test_age=si_age.fit_transform(X_test[['age']])
X_train_embarked=si_embarked.fit_transform(X_train[['embarked']])
X_test_embarked=si_embarked.fit_transform(X_test[['embarked']])
```

```python
X_train_age
```

```python
# Applying one hot encoding on sex and embarked
ohe_sex=OneHotEncoder(sparse_output=False,handle_unknown='ignore')
ohe_embarked=OneHotEncoder(sparse_output=False,handle_unknown='ignore')
X_train_sex=ohe_sex.fit_transform(X_train[['sex']])
X_train_embarked=ohe_embarked.fit_transform(X_train_embarked)
X_test_embarked=ohe_embarked.fit_transform(X_test_embarked)
X_test_sex=ohe_embarked.fit_transform(X_test[['sex']])
```

```python
X_train_sex
```

```python
X_train_rem=X_train.drop(columns=['sex','age','embarked'])
```

```python
X_test_rem=X_test.drop(columns=['sex','age','embarked'])
```

```python
X_train_transformed=np.concatenate((X_train_rem,X_train_age,X_train_sex,X_train_embarked),axis=1)
```

```python
X_test_transformed=np.concatenate((X_test_rem,X_test_age,X_test_sex,X_test_embarked),axis=1)
```

```python
clf=DecisionTreeClassifier()
clf.fit(X_train_transformed,y_train)
y_pred=clf.predict(X_test_transformed)
```

```python
y_pred
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```
