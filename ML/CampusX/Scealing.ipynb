```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv(r"D:\Data-Science\ML\CampusX\Social_Network_Ads.csv")
```

```python
df=df.iloc[:,2:]
```

```python
df.describe()
```

# Train Test Split

```python
from sklearn.model_selection import train_test_split
```

```python
X=df.iloc[:,0:-1]
y=df.iloc[:,-1]
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=0)
```

```python
X_train.shape ,X_test.shape,y_train.shape,y_test.shape
```

```python
X_train
```

# Standard Scaler

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler=StandardScaler()
```

```python
# Fit the scaler to the train set ,it will learn parameter
scaler.fit(X_train)
X_train_scaled=scaler.transform(X_train)
X_test_scaled=scaler.transform(X_test)
X_train_scaled
```

# It return Numpy array that why we have to convert it into dataframe

```python
X_train_scaled=pd.DataFrame(
    X_train_scaled,columns=X_train.columns
)
X_test_scaled=pd.DataFrame(
    X_test_scaled,columns=X_test.columns
)
```

```python
X_train_scaled
```

```python
np.round(X_train.describe())
```

```python
np.round(X_train_scaled.describe())
```

```python
fig, (ax1,ax2)=plt.subplots(ncols=2,figsize=(12,5))
ax1.scatter(X_train['Age'],X_train['EstimatedSalary'])
ax1.set_title('Before Scaling')
ax2.scatter(X_train_scaled['Age'],X_train_scaled['EstimatedSalary'],c='red')
ax2.set_title('After Scaling')
```

```python
fig,(ax1,ax2)=plt.subplots(ncols=2,figsize=(12,5))
#Before scealing
ax1.set_title('Before Scealing')
ax2.set_title('After scealing')
sns.kdeplot(X_train['Age'],ax=ax1)
sns.kdeplot(X_train['EstimatedSalary'],ax=ax1)
# After Sceaning
sns.kdeplot(X_train_scaled['Age'],ax=ax2)
sns.kdeplot(X_train_scaled['EstimatedSalary'],ax=ax2)
```

# Comparision of Distribution

```python
fig,(ax1,ax2)=plt.subplots(ncols=2,figsize=(12,5))
# Age Before scealing
ax1.set_title('before scaling')
ax2.set_title('after scaling')
sns.kdeplot(X_train['Age'],ax=ax1)
sns.kdeplot(X_train_scaled['Age'],ax=ax2)
```

# Why scaling is Important ?

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr=LogisticRegression()
```

```python
lr_scaled=LogisticRegression()
```

```python
lr.fit(X_train,y_train)
lr_scaled.fit(X_train_scaled,y_train)
```

```python
y_pred=lr.predict(X_test)
y_pred_scaled=lr_scaled.predict(X_test_scaled)
```

```python
from sklearn.metrics import accuracy_score
```

```python
print('Actual',accuracy_score(y_test,y_pred))
print('Scaled',accuracy_score(y_test,y_pred_scaled))
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
dt=DecisionTreeClassifier()
dt_scaled=DecisionTreeClassifier()
```

```python
dt.fit(X_train,y_train)
dt_scaled.fit(X_train_scaled,y_train)
```

```python
y_pred=dt.predict(X_test)
y_pred_scaled=dt_scaled.predict(X_test_scaled)
```

```python
print('Actual',accuracy_score(y_test,y_pred))
print('Scaled',accuracy_score(y_test,y_pred_scaled))
```
