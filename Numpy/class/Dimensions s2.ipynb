```python
# 3,2,3
l1=[10,20,30]      #k=3
l2=[l1,l1]         #n=2
l3=[l2,l2,l2]      #m=3
np.array(l3)       #m,n,k
```

```python
# 4,2,2,4
l1=[10,20,30,40]      #k=3
l2=[l1,l1]            #n=2
l3=[l2,l2]            #m=3
l4=[l3,l3,l3,l3]             
np.array(l4).shape    #m,n,k
```
