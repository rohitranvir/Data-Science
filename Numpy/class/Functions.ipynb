# 1️⃣ np.random.rand()
Meaning

Generates random floats

Range: 0.0 ≤ value < 1.0

Shape: (2,3)

```python
np.random.rand(2,2)
```

```python
np.random.random((2,2))
```

# 2️⃣ np.random.randn()
np.random.randn(2, 2)
Meaning
Generates random floats from a normal (Gaussian) distribution

Mean = 0, Standard deviation = 1

```python
np.random.randn(2,2)
```

# 3️⃣ np.random.uniform()
Meaning

Random floats between 10 and 20

Uniform distribution

Better than rand() when:

You need a specific range

```python
np.random.uniform(10,20,size=(2,2))
```

# 4️⃣ np.random.normal()
Meaning

Normal distribution

loc → mean

scale → standard deviation

Real use

Noise generation

ML weight initialization
| Parameter | Meaning                       |
| --------- | ----------------------------- |
| `loc`     | Mean (center of distribution) |
| `scale`   | Standard deviation (spread)   |
| `size`    | Shape of output array         |


```python
np.random.normal(loc=50,scale=10,size=(2,2))
```

```python
data=np.random.normal(5000,1,1000)
data.mean(),data.std()
```

```python
import matplotlib.pyplot as plt
```

```python
plt.hist(data)
plt.show()
```

# `8️⃣ np.random.seed()
Meaning

Fixes randomness

Makes results reproducible

Critical in:

ML experiments

Debugging

# Example without seed

```python
import numpy as np

print(np.random.randint(1, 10, 5))
print(np.random.randint(1, 10, 5))
```

# Example with seed

```python
np.random.seed(4)
print(np.random.randint(1,10,5))
print(np.random.randint(1,10,5))
```
