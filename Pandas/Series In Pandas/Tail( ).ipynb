# 8. PANDAS SERIES - TAIL() METHOD

The pandas.series.tail() method is used to return the last n rows of the Pandas
Series.

Let us now see an example to implement the tail() method in Python Pandas:

```python
import pandas as pd
data =[10,20,None,30,40]
df=pd.Series(data,index=["RowA","RowB","RowC","RowD","RowE"])
print(df.tail(1))
```
