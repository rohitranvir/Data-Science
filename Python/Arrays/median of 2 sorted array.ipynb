```python
def merg(nums1,nums2):
    li=[]
    for i in nums1:
        li.append(i)
    for j in nums2:
        li.append(j)
    li=sorted(li)
    if len(li)%2==1:
        left =0
        right=len(li)-1        
        mid=len(li)//2
        return float(li[mid])
    else:
        left =0
        right=len(li)-1  
        
        while left<right: 
            # print(left)
            # print(right)
            if right-left==1:
                mid=(li[left]+li[right])/2
                return float(mid)
            left+=1
            right-=1
nums1 = [1,2]
nums2 = [3,4]
print(merg(nums1,nums2))
```
