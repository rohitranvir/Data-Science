# File Handaling means creating ,reading ,writing  ,and modifying  file stored on disk.

# Common Modes
1. "r" => "read (File must be exist)"
2. "w" => "Write (Create or overwrite file)"
3. "a" => "append(adds at the end)"
4. "x" => "Create (fails if file exists)"
5. "rb","wb"=> "Binary Files(Image models)"\

# Risk :
Using "w" can permantely delete existing data

```python
from PIL import Image

img1 = Image.open(r"C:\Users\rohit\OneDrive\Pictures\Smiley.jpg")
img2 = Image.open(r"C:\Users\rohit\OneDrive\Pictures\smi.jpg")

# ensure same width
width = max(img1.width, img2.width)
height = img1.height + img2.height

new_img = Image.new("RGB", (width, height))

new_img.paste(img1, (0, 0))
new_img.paste(img2, (0, img1.height))

new_img.save("combined_vertical.jpg")
```
