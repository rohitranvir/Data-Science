# Wpp which will accept list of numerical values  and find sum of possitive numerical values and negative values by using special function in python


```python
s=[float(val) for val in input("Enter a words which seperated by space").split(",")]
pos=list(filter(lambda x:x>0,s))
print(pos)
neg=list(filter(lambda x:x<0,s))
print(neg)
posres=functools.reduce(lambda x,y:x+y,pos)
negres=functools.reduce(lambda x,y:x+y,neg)

print("positive ==> ",posres)
print("negative ==> ",negres)
```

# Wpp which will accept list of  employee postitive salary.
    1. obtain those employee salary which are rangeing from 0 to 500 and give 10% hike. give 20% hike to those 
       emoployees whose salary rangine 501 to 1000.print the total salary of 10% hiked employees and 20% hiked employees
       (maximum salary is 1000 and minsal is 0)
    

```python
s=[float(val) for val in input("Enter a words which seperated by space").split()]
under5=list(filter(lambda x:x if x>0 and x<=500 else 0,s))
print(under5)
more5=list(filter(lambda x:x if x>500 and x<=1000 else 0,s))
print(more5)
posres=list(map(lambda x:x+x*(10/100),under5))
negres=list(map(lambda x:x+x*(20/100),more5))
print("positive ==> ",posres)
print("negative ==> ",negres)
psum=functools.reduce(lambda x,y:x+y,posres)
nsum=functools.reduce(lambda x,y:x+y,negres)
print("sumof possitive ",psum)
print("sumof possitive ",nsum)
```
