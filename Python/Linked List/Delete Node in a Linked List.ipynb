```python
class node:
    def __init__(self,data):
        self.data=data
        self.next=None

l=[4,5,1,9]
head=node(l[0])
curr=head
for i in l[1:]:
    curr.next=node(i)
    curr=curr.next
```

```python
val=5
def delnod(head,val):
    curr=head
    while curr.next.data!=val:
        curr=curr.next
    curr.next=curr.next.next
delnod(head,val)
```
