```python
class Node:
    def __init__(self,data):
        self.data=data
        self.next=None
head=Node(10)
head.next=Node(20)
head.next.next=Node(30)
head.next.next.next=Node(40)
head.next.next.next.next=Node(50)
head.next.next.next.next.data
```

```python
def traverse(head):
    print(head.data)
    curr=head
    while curr.next!=None:
        curr=curr.next
        print(curr.data)
traverse(head)
```

```python
# delete first node
head=head.next
traverse(head)
```

```python
# delete kth node
curr=head
k=2
for i in range(k-1):
    curr=curr.next
curr.next=curr.next.next
traverse(head)
```
