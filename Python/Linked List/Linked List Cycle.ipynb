```python
class node:
    def __init__(self,data):
        self.data=data
        self.next=None
l= [3,2,-4,0,-4]
head=node(l[0])
curr=head
for i in l[1:]:
    curr.next=node(i)
    curr=curr.next
```

```python
def cycle(head):
    curr=head
    myset=set()
    while curr is not None:        
        if curr in myset:
            return True
        myset.add(curr)
        curr=curr.next
    return False
cycle(head)
```

```python
def trav(head):
    curr=head
    while curr != None:
        print(curr.data)
        curr=curr.next
trav(head)
```
