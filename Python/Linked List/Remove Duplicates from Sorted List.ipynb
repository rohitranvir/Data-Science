```python
class node:
    def __init__(self,data):
        self.data=data
        self.next=None


l= [1,1,2]
head=node(l[0])
head=node(l[0])
curr=head
for i in l[1:]:
    curr.next=node(i)
    curr=curr.next
```

```python
# removing duplicate node
def dup(head):
    curr=head
    while curr.next != None:
        if curr.data==curr.next.data:
            curr.next=curr.next.next
        curr=curr.next
dup(head)    
```

```python
def trav(head):
    curr=head
    l=[]
    while curr!=None:
        l.append(curr.data)
        curr=curr.next   
    di={}
    for i in range(len(l)):
        if l[i] not in di :
            di[l[i]]=1
        else:
            di[l[i]]+=1
    print(di)            
trav(head)
```
