```python
class node:
    def __init__(self,data):
        self.data=data
        self.next=None
    
    
l = [1,2,3,4,5]
n = 2
head=node(l[0])
curr=head
for i in l[1:]:
    curr.next=node(i)
    curr=curr.next
```

```python
def delete(head,val):
    p1=head
    p2=head
    for i in range(val):
        p2=p2.next
    while p2.next!=None:
        p1=p1.next
        p2=p2.next
    p1.next=p1.next.next
delete(head,n)
        
```

```python
def trav(head):
    curr=head
    while curr != None:
        print(curr.data)
        curr=curr.next
trav(head)
```
