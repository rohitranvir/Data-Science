```python
class Node:
    def __init__(self,data):
        self.data=data
        self.next=None
head=Node(10)
head.next=Node(20)
head.next.next=Node(30)
head.next.next.next=Node(40)
head.next.next.next.next=Node(50)
head.next.next.next.next.data
```

```python
def traverse(head):
    print(head.data)
    curr=head
    while curr.next!=None:
        curr=curr.next
        print(curr.data)
traverse(head)
```

```python
fnode=Node(2)
fnode.next=head
head=fnode
```

```python
fnode=Node(6)
def adlast(head):
    curr=head
    while curr.next!=None:
        curr=curr.next
    curr.next=fnode
adlast(head)
```

```python
kthnode=Node(5)
k=3
curr=head
for i in range(k-1):
    curr=curr.next
kthnode.next=curr.next
curr.next=kthnode
```
