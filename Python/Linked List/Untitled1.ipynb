```python
class node:
    def __init__(self,data):
        self.data=data
        self.next=None
head=node(10)
head.next=node(20)
head.next.next=node(30)
print(head.data)
print(head.next.data)
print(head.next.next.data)
```

```python
def traverse(head):
    curr=head
    while curr!=None:
        print(curr.data)
        curr=curr.next
traverse(head)
```

```python
firstnode=node(50)
firstnode.next=head
head=firstnode
```

```python
lstnode=node(60)
curr=head
while curr.next!=None:
    curr=curr.next
curr.next=lstnode
traverse(head)    
```

```python
newnode=node(5)
curr=head
k=2
for i in range(k-1):
    curr=curr.next
newnode.next=curr.next
curr.next=newnode
traverse(head)    
```
