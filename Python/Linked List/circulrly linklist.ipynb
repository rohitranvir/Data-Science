```python
class Node:
    def __init__(self,data):
        self.data=data
        self.next=None
head=Node(10)
head.next=Node(20)
head.next.next=Node(30)
head.next.next.next=Node(40)
head.next.next.next.next=Node(50)
head.next.next.next.next.next=head
if head is None:
    print("List is empty")
else:
    curr = head
    while True:
        print(curr.data,end=" ")
        curr = curr.next
        if curr == head:
            break
```
