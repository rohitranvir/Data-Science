```python
class node:
    def __init__(self,data):
        self.data=data
        self.next=None
    
l=[1,2,3,4,5]
head=node(l[0])
curr=head
for i in l[1:]:
    curr.next=node(i)
    curr=curr.next
```

```python
def tranverse(head):
    curr=head
    stack=[]
    while curr!=None:
        stack.append(curr.data)
        curr=curr.next
    curr=head
    while curr.next != None:
        e=stack.pop()
        curr.data=e
        curr=curr.next
tranverse(head)
```

```python
def rev(head):
    prev=None
    curr=head
    nxt=None
    while curr.next!= None:
        nxt=curr.next
        curr.next=prev
        prev=curr
        curr=nxt
rev(head)        
    
```
