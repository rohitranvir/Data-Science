```python
class Node:
    def __init__(self,data):
        self.data=data
        self.next=None
head=Node(10)
head.next=Node(20)
head.next.next=Node(30)
head.next.next.next=Node(40)
head.next.next.next.next=Node(50)
print(head.next.next.data)
```

```python
def tranverse(head):
    curr=head
    while curr!=None:
        print(curr.data)
        curr=curr.next
tranverse(head)
```

```python
# Insertation at beggining
newnode=Node(5)
newnode.next=head
head=newnode
```

```python
# Insertation at the end
newnode=Node(1)
curr=head
while curr.next!=None:
    curr=curr.next
    # print(curr.data)
    
curr.next=newnode
tranverse(head)
```

```python
# traverse 
def traverse(head):
    curr=head
    while curr.next!=None:
        print(curr.data)
        curr=curr.next
traverse(head)
```
