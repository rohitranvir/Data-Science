```python
# max_sum
def max_sum(arr,t):
    msum=float("-inf")
    left=0
    right=t-1
    while right<len(arr):
        temp=0
        for i in range(left,right+1):
            temp+=arr[i]
        if temp>msum:
            msum=temp
        left+=1
        right+=1

    return msum
            
        
    
max_sum([-1,2,3,3,4,5,-1],4)
```
