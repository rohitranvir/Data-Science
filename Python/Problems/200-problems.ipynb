```python
def longe(s):
    res=0
    dic={}
    left=0
    right=0
    while right<len(s):
        if s[right] not in dic:
            dic[s[right]]=1
        else:
            # if len(res)<(len(s[right])-len(s[left])):
            #     res=s[left:right:1]
            res=max(res,right-left)
            left+=1
        right+=1
    return res
longe('abcabcbb')
```
