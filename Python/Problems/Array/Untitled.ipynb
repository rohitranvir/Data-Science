```python
def sortornot(l):
    res=True
    for i in range(len(l)-1):
        if l[i]>l[i+1]:
            return False
    return res
l=[1,2,3,4,5]
sortornot(l)
```

```python
def smlslg(l):
    sm=float("inf")
    ssm=float("inf")
    lg=float("-inf")
    slg=float("-inf")
    for i in range(len(l)):
        if l[i]<sm:        
            ssm=sm
            sm=l[i]
        elif l[i]<ssm and l[i]!=sm:
            ssm=l[i]
    for i in range(len(l)):
        if l[i]>lg:
            slg=lg
            lg=l[i]
        elif l[i]>slg and l[i]!=lg:
            slg=l[i]
    return ssm,slg
l=[2,5,4,1,2]  
smlslg(l)
```
