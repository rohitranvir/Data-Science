#  TC= O(n(n+1)/2)==O(n**2)

### Brute force

```python
p = [7,6,4,3,1]
max_p=0
for b in range(len(p)-1):
    for s in range(b+1,len(p)):  
        if p[s]>p[b]:
            t1=p[s]-p[b]
            if max_p<t1:
                max_p=t1
print(max_p)
                
```

# Optimal Solution

```python
p=[7,1,5,3,6,4]
max_p=0
min_p=float('inf')
for i in range(0,len(p)):
    min_p=min(min_p,p[i])
    max_p=max(max_p,p[i]-min_p)
print(max_p)
```

```python
p=[7,1,5,3,6,4]
min=1e5
profit=0
for i in p:
    if i<min:
        min=i
    if i-min>profit:
        profit=i-min
print(profit)
```
