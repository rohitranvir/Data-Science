```python
def combsum(nums,target):
    res=[]
    for i in range(len(nums)):
        left =i
        right=len(nums)-1        
        while left<=right:
            total=nums[i]+nums[left]+nums[right]
            if total==target:
                res.append([nums[i],nums[left],nums[right]])
                left+=1
                # right-=1
            # elif total>target:
            #     right-=1
            else:
                left+=1
    return res        
nums=[2,3,6,7]
target = 7
combsum(nums,target)
```
