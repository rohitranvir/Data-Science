# Brute Force Method   TC=O(n**2)   , sc=O(1)

```python
n = [0,3,7,2,5,8,4,6,0,1]
n.sort()
```

```python
max_c=0
for i in range(0,len(n)):
    num=n[i]
    count=1
    while num+1 in n:
        count+=1
        num+=1
    max_c=max(count,max_c)
print(max_c)        
```

```python
n = [0,3,7,2,5,8,4,6,0,1]
s=set(n)
long=0
for i in s:
    if i-1 not in s:
        nums=i
        count=1
        while nums+1 in s:
            nums+=1
            count+=1
    long=max(count,long)
print(long)
```

# better Solution

```python
n = [1,0,1,2]
n.sort()
min_i=float('-inf')
max_c=0
count=0
for i in range(len(n)):
    if n[i]-1==min_i:
        count+=1
        min_i=n[i]
    elif n[i]!=min_i:
        count=1
        min_i=n[i]
    max_c=max(max_c,count)
print(max_c)
```
