```python
def lowb(arr,target):
    n=len(arr)
    low=0
    high=n-1
    lb=-1
    while low<=high:
        mid=(low+high)//2
        if target<=arr[mid]:
            high=mid-1
            lb=mid
        else:
            low=mid+1
            # lb=arr[low]
    return lb
arr = [1, 2, 4, 4, 4, 4,6, 7]
target = 4
lowb(arr,target)
```
