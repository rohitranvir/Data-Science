# Optimal Solution 
#### TC= O(n*n)+O(n*n)
#### SC=O(1)

```python
def rot(m):
    n=len(m)
    for i in range(n):
        for j in range(i+1,n):
            if i!=j:
                m[i][j],m[j][i]=m[j][i],m[i][j]
    for i in m:
        print(i)
m = [[1,2,3],
    [4,5,6],
    [7,8,9]
    ]
rot(m)
```
