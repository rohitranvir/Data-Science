# Brute force 

### TC= O((n+m)*(n*m))+O(N*M)

```python
def mat(m,row,col):
    for i in range(len(m)):
        if m[i][col]!=0:
            m[i][col]=float('inf')
    for j in range(len(m[0])):
        if m[row][j]!=0:
            m[row][j]=float('inf')
    return m
def setzero(m):
    
    for i in range(len(m)):
        for j in range(len(m[0])):
            if m[i][j]==0:
                mat(m,i,j)
    for i in range(len(m)):
        for j in range(len(m[0])):
            if m[i][j]==float('inf'):
                m[i][j]=0
    return m
    
m=[[1,1,1],
   [1,0,0],
   [1,1,1]
  ]
setzero(m)
for i in m:
    print(i)
```

```python
class Solution:
    def mat(self,m,col,row):
        for i in range(len(m)):
            if m[col][i] !=0:
                m[col][i]=float('inf')
        for j in range(len(m[0])):
            if m[j][row]!=0:
                m[j][row]=float('inf')
        return m         

    def setZeroes(self, m):
        
        for i in range(len(m)):
            for j in range(len(m[0])):
                if m[i][j]==0:
                    self.mat(m,i,j)
        for i in range(len(m)):
            for j in range(len(m[0])):
                if m[i][j]==float('inf'):
                    m[i][j]=0
        return m
s=Solution()
m=[[1,1,1],
[1,1,0],
[1,1,1]
]
s.setZeroes(m)
for i in m:
    print(i)
```
