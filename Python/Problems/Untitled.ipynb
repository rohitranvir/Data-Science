```python
def getSecondLargest(ar):
    larg=float('-inf')
    sec_l=float('-inf')
    for i in ar:
        if i>larg:
            sec_l=larg
            larg=i
        elif i>sec_l and i!=larg:
            sec_l=i    
    return sec_l if sec_l !=float('-inf') else -1
    
l=[10 ,10 ,10]
getSecondLargest(l)
```

```python
def getSecondOrderElements(a):
    larg=float('-inf')
    sec_l=float('-inf')
    for i in a:
        if i>larg:
            sec_l=larg
            larg=i
        elif i>sec_l and i!=larg:
            sec_l=i
    return sec_l if sec_l !=float('-inf') else -1
l=[10 ,10 ,10]
getSecondOrderElements(l)
```

```python
def repeatchr(s):
    max_len=0
    for i in range(len(s)):
        dic=[]
        long=0
        for j in range(i,len(s)):
            if s[j] in dic:
                break
            else:
                long+=1
                dic.append(s[j])
        max_len=max(max_len,long)
    return max_len
s = "aaaa"
repeatchr(s)
```
