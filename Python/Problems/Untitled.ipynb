```python
def getSecondLargest(ar):
    larg=float('-inf')
    sec_l=float('-inf')
    for i in ar:
        if i>larg:
            sec_l=larg
            larg=i
        elif i>sec_l and i!=larg:
            sec_l=i    
    return sec_l if sec_l !=float('-inf') else -1
    
l=[10 ,10 ,10]
getSecondLargest(l)
```

```python
def getSecondOrderElements(a):
    larg=float('-inf')
    sec_l=float('-inf')
    for i in a:
        if i>larg:
            sec_l=larg
            larg=i
        elif i>sec_l and i!=larg:
            sec_l=i
    return sec_l if sec_l !=float('-inf') else -1
l=[10 ,10 ,10]
getSecondOrderElements(l)
```
