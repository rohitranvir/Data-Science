```python
def countleseq(a,b):
    res=[]
    for i in a:
        count=0
        for j in b:
            if j<=i:
                count+=1
        res.append(count)
    return res
a= [1, 2, 3, 4, 7, 9]
b= [0, 1, 2, 1, 1, 4]
countleseq(a,b)
```
