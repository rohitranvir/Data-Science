```python
def commonpre(strs):
    res=""
    l=min(strs,key=len)
    for i in range(len(l)):
        c=strs[0][i]
        for j in strs:
            if j[i]!=c:
               return res
        res+=c
    return res
strs = ["flower","flow","flight"]
commonpre(strs)
```
