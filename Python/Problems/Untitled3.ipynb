```python
def merge(nums1,m,nums2, n):
    if m==0 and n==1:
        c=n-1
        nums1[c]=nums2[0]
    
    n1=m-1
    n2=n-1
    l=len(nums1)-1
    while n1>=0 and n2>=0:
        if nums1[n1]>nums2[n2]:
            nums1[l]=nums1[n1]
            n1-=1
            l-=1
        else:
            nums1[l]=nums2[n2]
            n2-=1
            l-=1
    
    while n2 >= 0:
        nums1[l] = nums2[n2]
        n2 -= 1
        n1 -= 1
    return nums1
            
    
nums1 = [2,0]
m=1
nums2 = [1]
n=1
merge(nums1,m,nums2, n)
```
