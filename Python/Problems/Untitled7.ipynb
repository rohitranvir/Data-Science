# Take a 3-digit number and check if all digits are distinct.

```python
def bet(c):
    res={}
    while c>0:
        temp=c%10
        if temp in res:
            res[temp]+=1
        else:
            res[temp]=1
        c=c//10
    for i,j in res.items():
        if j>1:
            return "Not valid"    
    return "Valid"
bet(1233)
```
