```python
def getFloorAndCeil(a, n, x):
    floor=-1
    ceil=-1
    low=0
    high=len(a)-1
    while low<=high:
        mid=(low+high)//2
        if a[mid]==x:
            return [a[mid],a[mid]]
        elif a[mid]>=x:
            ceil=a[mid]
            high=mid-1
        else:
            floor=a[mid]
            low=mid+1
    return [floor,ceil]

n=6 
x=5
a=[3, 4, 7, 8, 8, 10]   
getFloorAndCeil(a, n, x)   
```
