# Brute force solution

```python
def searchRange(nums, target):
    first=-1
    last=-1
    
    for i in range(len(nums)):
        if nums[i]==target:
            first=i
            for j in range(i+1,len(nums)):
                if nums[j]==target:
                    last=j
            break
    return [first,last]        
    
nums = [5,7,7,8,8,10]
target = 10
searchRange(nums, target)
```
