```python
def freq(s):
    res={}
    m=0
    for i in s:
        if i in res:
            res[i]+=1
        else:
            res[i]=1
    for i in res.values():
        if i>m:
            m=i
    for key,val in res.items():
        if val==m:
            return [key]
    
s="aabbbbccdd"
freq(s)
```
