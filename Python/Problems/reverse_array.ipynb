```python
def rev(arr):
    # left =0
    # right=len(arr)-1
    maxe=arr[0]
    mine=arr[0]
    for i in arr:
        if i<mine:
            mine=i
        elif i>maxe:
            maxe=i        
    return maxe,mine
rev([1,2,3,4,5,6])
```

```python
def move_zeros(arr):
    count=0
    dic={}
    for i in range(len(arr)):
        if arr[i]==0:
            arr[i]=arr[i+1]
            # arr[j+1]=arr[j+2]    
            count+=1
        if arr[i] in dic:
            dic[arr[i]]+=1
        else:
            dic[arr[i]]=1
    dic=[k for k,v in dic.items() if v<=2]    
    for i in range(count):
        dic.append(0)
    return dic
print(move_zeros([0,1,0,3,12]))
```

```python
def move_zeros(arr):
    count=0
    for i in range(len(arr)):
        if arr[i]!=0:
            arr[count]=arr[i]
            count+=1
    while count<len(arr):
        arr[count]=0
        count+=1
    return arr
print(move_zeros([0,1,0,3,12]))
```
