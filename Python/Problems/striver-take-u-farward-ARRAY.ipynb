```python
def larg(arr):
    secl=-1
    larg=arr[0]
    for i in range(len(arr)):
        if arr[i]>larg:
            secl=larg
            larg=arr[i]
        elif arr[i]>secl and arr[i]<larg:
            secl=arr[i]
    return secl
arr=[9,2,3,7,4,5,6,6,8,4,8,]
larg(arr)
```

```python
def sort(arr):
    curr=arr[0]
    for i in range(len(arr)):
        if arr[i]>=curr:
            curr=arr[i]
        else:
            return False
    return True
arr=[1,2,1,3,3,4]
sort(arr)
```
