# Longest repeating charactor Replacement

```python
def repeat(s,k):
    maxlen=0
    for i in range(len(s)):
        dic={}
        count=0
        for j in range(i,len(s)):                
            if s[j] not in dic:
                dic[s[j]]=1
            else:
                dic[s[j]]+=1
            count=max(count,dic[s[j]])
            changes=(j-i+1)-count
            if changes<=k:
                maxlen=max(maxlen,j-i+1)   
            else:
                break
    return maxlen

s='aababba'
repeat(s,2)
```

```python
def atmost(nums,k):
    count=0
    for i in range(len(nums)):
        suma=0
        for j in range(i,len(nums)):
            suma+=nums[j]
            if suma==k:
                count+=1
    return count

nums = [1, 1, 0, 1, 0, 0, 1]
k = 2
atmost(nums,k)
```

```python
def atmost(nums,k):
    if k<0:
        return 0
    count=0
    suma=0
    left=0
    right=0
    while right<len(nums):
        suma+=nums[right]
        while suma>k:
            suma-=nums[left]
            left+=1
        count+=(right-left+1)
        right+=1
    return count
def binsub(nums,k):
    return atmost(nums, k) - atmost(nums, k - 1)
nums = [1, 1, 0, 1, 0, 0, 1]
k = 2
binsub(nums,k)
```

# Set matrix to zero

```python
def setzero(matrix):
    row=set()
    col=set()
    for i in range(len(matrix)):
        for j in range(len(matrix[0])):
            if matrix[i][j]==0:
                row.add(i)
                col.add(j)
    for i in range(len(matrix)):
        for j in range(len(matrix[0])):
            if i in row or j in col:
                matrix[i][j]=0
    return matrix
matrix=[
        [0,1,2,0],
        [3,4,5,2],
        [1,3,1,5]
        ]
setzero(matrix)
```

# Longest Substring Without Repeating Characters


```python
def substr(s):
    res=0
    left=0
    right=0
    dic={}
    
    while right < len(s):
        if s[right] not in dic:
            dic[s[right]]=1
        else:
            dic[s[right]]+=1
        while dic[s[right]]>1:
            dic[s[left]] -=1
            left+=1
        res=max(res,(right-left)+1)
        right+=1
    return res
s="abcabcbb"
substr(s)
```

# Longest Common Prefix

```python
def prefix(strs):
    res=""
    for i in range(len(strs[0])):
        temp=strs[0][i]
        for j in range(1,len(strs)):
            if i>=len(strs[j]) or strs[j][i]!=temp:
                return res
        res+=temp              
    return res            
strs = ["flower", "flow", "flight"]
prefix(strs)
```

```python
def minsub(s, t):
    res = ""

    for i in range(len(s)):
        dic = {}
        for ch in t:
            if ch in dic:
                dic[ch] += 1
            else:
                dic[ch] = 1

        count = len(t)

        for j in range(i, len(s)):
            if s[j] in dic:
                if dic[s[j]] > 0:
                    count -= 1
                dic[s[j]] -= 1

            if count == 0:
                temp = s[i:j+1]
                if res == "" or len(temp) < len(res):
                    res = temp
                break   # no need to expand more

    return res
```

# Group of Annagram

```python
def anna(inp):
    dic={}
    for i in inp:
        key="".join(sorted(i))
        # print(key)
        if key in dic:
            dic[key].append(i)
        else:
            dic[key]=[i]
    return list(dic.values())
            
Input= ["eat","tea","tan","ate","nat","bat"]
anna(Input)
```
