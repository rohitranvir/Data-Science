# Longest repeating charactor Replacement

```python
def repeat(s,k):
    maxlen=0
    for i in range(len(s)):
        dic={}
        count=0
        for j in range(i,len(s)):                
            if s[j] not in dic:
                dic[s[j]]=1
            else:
                dic[s[j]]+=1
            count=max(count,dic[s[j]])
            changes=(j-i+1)-count
            if changes<=k:
                maxlen=max(maxlen,j-i+1)   
            else:
                break
    return maxlen

s='aababba'
repeat(s,2)
```

```python
def repeat(s,k):
    maxlen=0
    dic={}
    left=0
    right=0  
    maxf=0
    while right <len(s):
        if s[right] not in dic:
            dic[s[right]]=1
        else:
            dic[s[right]]+=1
        maxf=max(maxf,dic[s[right]])
        while ((right-left+1)-maxf>k):
            dic[s[left]]-=1
            left+=1

        if ((right-left+1)-maxf)<=k:
            maxlen=max(maxlen,right-left+1)
        right+=1
    return maxlen
s='aababba'
repeat(s,2)
```

#  Binary sub array with sum k

```python
def atmost(nums,k):
    count=0
    for i in range(len(nums)):
        suma=0
        for j in range(i,len(nums)):
            suma+=nums[j]
            if suma==k:
                count+=1
    return count

nums = [1, 1, 0, 1, 0, 0, 1]
k = 2
atmost(nums,k)
```

```python
def atmost(nums,k):
    if k<0:
        return 0
    count=0
    suma=0
    left=0
    right=0
    while right<len(nums):
        suma+=nums[right]
        while suma>k:
            suma-=nums[left]
            left+=1
        count+=(right-left+1)
        right+=1
    return count
def binsub(nums,k):
    return atmost(nums, k) - atmost(nums, k - 1)
nums = [1, 1, 0, 1, 0, 0, 1]
k = 2
binsub(nums,k)
```

# count Number of Nice Subarrays

```python
def nice(nums,k):
    count=0
    for i in range(len(nums)):
        odd=0
        suma=0
        for j in range(i,len(nums)):
            suma+=nums[j]
            if nums[j]%2!=0:
                odd+=1
            if odd==k:
                count+=1
                break
    return count
k=3
nums=[1,1,2,1,1]
nice(nums,k)
```
