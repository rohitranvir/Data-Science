```python
def fruit(nums):
    res=[]
    maxlen=0    
    for i in range(len(nums)):
        s=set()
        for j in range(i,len(nums)):
            s.add(nums[j])
            if len(s)<=2:
                maxlen=max(maxlen,j-i+1)
        res.append(s)
    return maxlen

        
nums=[3,3,3,1,2,1,1,2,3,3,4,3,3]
fruit(nums)
```
