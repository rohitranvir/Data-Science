# Matching a sequence of 10 digit
- Here we are focusing on to identify
    - \d       : one digit
    - \d+      : More than one digit
    - \d{3}    : Exactly 3 digit
    - \d{2,4}  : digit between 2 to 4 digit
    - \d{3,}   : Atleast 3 digit


```python
import re 
pattern = '\d{3}'
str1='my mobile number is 9158000676'
re.findall(pattern,str1)
```

```python
import re 
pattern = '\d{2}'
str1='my mobile number is 9158000676'
re.findall(pattern,str1)
```

```python
import re 
pattern = '\d{3,}'
str1='my mobile number is 9158000676 4546 554 21'
re.findall(pattern,str1)
```

```python
import re
str1="my name is 123-345-6540"
pattern=r"\d{3,}-\d{3,}-\d{3,}"
re.findall(pattern,str1)
```

```python
import re
str1="My name is (123)-456-7890"
pattern=r"\(\d{3,}\)-\d{3,}-\d{3,}"
re.findall(pattern,str1)
```

```python
import re
str1="My name is +91-123-456-7890"
pattern=r"\+\d{2,}-\d{3,}-\d{3,}-\d{4,}"
re.findall(pattern,str1)
```

# w and W family

- \w : a-z and A-Z and 0-9 and _
   - single charactor
   - ex: str1= "Ram" : r,a,m
   - \w+ : more than one charactor
     
       - ex: str1 ="Ram" : Ram
  

   - \W: Single punchucation
   - \W+ : Multiple Panchucation

```python
s1='Ramesh 25'
s2='My name is Ramesh'
s3="Hello@123"
s4="Hello#!world@123"
```

```python
import re
pt=r"\w+"
re.findall(pt,s1)
```

```python
import re
pt=r"\w+"
re.findall(pt,s2)
```

```python
import re
pt=r"\w+\W\w+"
re.findall(pt,s3)
```

```python
import re
pt=r"\w+\W+\w+\W+\w+\d+"
re.findall(pt,s4)
```

# s and S Family

- \s : identify space tab and new line
    - Single space only
- \s+ : identify multiple spaces , tabs  and new lines

- \S  : non space charactors single
   - ex: s1="Hi A"=H i and A
- \S+ : non space charactor word will come

```python
s1="good morning"
s2="good  morning"
s3="Hi A"
s4="my name is Ramesh"
```

```python
import re
pt=r"\S+"
re.findall(pt,s4)
```

```python
import re
pt=r"\S+"
re.findall(pt,s2)
```

# Symbols

- Caret  : start of string
- Dollar : end of string

```python
s1="good morning"
s2="good  morning"
s3="Hi A"
s4="9158000676"
```

```python
import re
pt=r"morning$"
re.findall(pt,s1)
```

```python
import re
pt=r"\d{3}$"
re.findall(pt,s4)
```
