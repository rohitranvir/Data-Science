```python
data= [11,10,12,14,12, 15, 14, 13, 15, 102, 12, 14, 17, 19, 107, 10, 13, 12, 14, 12, 108, 12, 11, 14, 13, 15, 10, 15, 12, 10, 14, 13, 15, 10]
```

```python
# Outliers
plt.boxplot(data)
```

```python
#Z-score

def detect_outliers(data):
    outliers=[]
    threshold=3 # 3rd standard deviravtion
    mean=np.mean(data)
    std=np.std(data)
    for i in data:
        z_score=(i-mean)/std
        if np.abs(z_score)>threshold:
            outliers.append(i)
    return outliers   
```

```python
detect_outliers(data) #this are the outliers
```
