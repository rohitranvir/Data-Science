# Standart Deviation
The standard deviation is a measure of
the amount of variation or dispersion
of a set of values. A low standard
deviation indicates that the values
tend to be close to the mean (also
called the expected value) of the set,
while a high standard deviation
indicates that the values are spread
out over a wider range.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=sns.load_dataset('titanic')
mn=df['age'].mean()
```

```python
df.head(3)
```

```python
df['age'].var()
```

```python
df['age'].std()
```

```python
sns.histplot(x='age',data=df)
plt.plot([mn for i in range(0,100)],[i for i in range(0,100)],c='yellow',label="mode")
plt.show()
```
