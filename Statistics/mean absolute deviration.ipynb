```python
a=np.array([75,65,73,68,72,76])
b=np.array([90,97,43,96,93,51])
no=np.array([1,2,3,4,5,6])
mean=np.mean(a)
```

```python
mean1=np.mean(a)
mean2=np.mean(b)
```

```python
mad_a=np.mean(np.abs(a-mean1))
mad_b=np.mean(np.abs(b-mean2))
```

```python
mad_a,mad_b
```

```python

plt.figure(figsize=(6,3))

plt.scatter(a,no,c='red',label='Section A')
plt.scatter(b,no,c='green',label='Section B')
plt.plot([70,70,70,70,70,70],no,c='blue',label='mean')
plt.legend()
plt.show()
```
