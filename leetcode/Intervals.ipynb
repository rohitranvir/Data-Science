```python
def mergint(inter):
    inter.sort()
    s=0
    e=1
    i=1
    for _ in range(1,len(inter)):
        if i>=len(inter):
            break
        # print(inter[i+1][e])
        if inter[i][s]<= inter[i-1][e]:
            inter[i-1][s]=min((inter[i][s]),inter[i-1][s])
            inter[i-1][e]=max((inter[i][e]),inter[i-1][e])
            inter.pop(i)
        else:
            i=i+1
    return inter
    # [[1,3],[2,6],[8,10],[15,18]]
    # [[1,4],[0,2],[3,5]]
    # [[1,4],[4,5]]
inter= [[1,4],[0,2],[3,5]]
mergint(inter)
```
