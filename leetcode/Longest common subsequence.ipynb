```python
def longestCommonSubsequence(text1, text2):
    count=0
    temp=text2
    for i in text1:
        for j in range(len(temp)):
            if i == temp[j]:
                count+=1
                temp=text2[j+1:]
                break
    return count
text1 = "oxcpqrsvwf"
text2 = "shmtulqrypy"
longestCommonSubsequence(text1, text2)
```

```python
def longestCommonSubsequence(text1, text2):
    # count=0
    m=len(text2)
    n=len(text1)
    dp=[[0]*(m+1) for _ in range(n+1)]
    for i in range(n):
        for j in range(m):
            if text1[i]==text2[j]:
                dp[i+1][j+1]=dp[i][j]+1
            else:
                dp[i+1][j+1]=max(dp[i+1][j],dp[i][j+1])
    return dp[n][m]
            
text1 = "oxcpqrsvwf"
text2 = "shmtulqrypy"
longestCommonSubsequence(text1, text2)
```
