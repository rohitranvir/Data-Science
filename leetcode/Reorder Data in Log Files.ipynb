```python
def reorderlogfile(logs):
    res1,res2=[],[]
    for log in logs:
        if (log.split()[1].isdigit()):
            res2.append(log)
        else:
            res1.append(log.split())
    # print(res1)
    # print(res2)
    res1.sort(key=lambda x:(x.split()[1], x.split()[0]))
    return res1+res2
    
logs = ["dig1 8 1 5 1","let1 art can","dig2 3 6","let2 own kit dig","let3 art zero"]
reorderlogfile(logs)
```
