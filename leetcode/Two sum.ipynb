```python
class Solution(object):
    def longestCommonPrefix(self, s):
        if len(s)==0:
            return ""
        res=""
        for i in range(len(s[0])):
            for j in s:
                if i>=len(j) or j[i] !=s[0][i]:
                    return res
            res+=s[0][i]
        return res
s=Solution()
strs = ["dog","racecar","car"]
s.longestCommonPrefix(strs)
```
