```python
def binsearch(arr,target):
    n=len(arr)
    low=0 
    high=n-1
    while low<high:
        mid=(low+high)//2
        if arr[mid]==target:
            return mid
            
        if target<arr[mid]:
            high=mid-1
        else:
            low=mid+1
        
l=[2, 5, 8, 12, 16, 23, 38, 56, 72, 91]
target = 23
binsearch(l,target)
```

```python
def binsearch(arr,low,high,target):    
    if low>high:
        return -1 
    mid=(low+high)//2
    if target==arr[mid]:
        return mid
    if target>arr[mid]:
        low=mid+1
        return binsearch(arr,low,high,target)
    elif target<arr[mid]:
        high=mid-1
        return binsearch(arr,low,high,target)
        
        
        
        
l=[2, 5, 8, 12, 16, 23, 38, 56, 72, 91]
target = 23
low=0
high=len(l)-1
binsearch(l,low,high,target)
```
