```python
from collections import Counter
def ana(s,p):
    if len(s)<len(p):
        return []
    res=[]
    k=len(p)
    
    p_count=Counter(p)
    window=Counter(s[:k])
    if p_count==window:
        res.append(0)
    print(window)
    print(p_count)   
        
s = "cbaebabacd"
p = "abc"
ana(s,p)
```
