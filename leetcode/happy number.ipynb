```python
class Solution(object):
    def isHappy(self, n):
        seen=set()        
        while n!=1:
            if n in seen:
                return False
            else:
                seen.add(n)
            res=0
            while n>0:
                t1=n%10
                res+=t1**2
                n=n//10
            n=res
        return True
        
s=Solution()
n=1
s.isHappy(n)
        
```
