```python
class into:
    def rom(self,num):
        dic={
            1000:"M",
            900:"CM",
            500:"D",
            400:"CD",
            100:"C",
            90:"XC",
            50:"L",
            40:"XL",
            10:"X",
            9:"IX",
            5:"V",
            4:"IV",
            1:"I",    
            }
        s=""
        for i in dic:
            while num>=i:
                s+=dic[i]
                num-=i
        return s
s=into()
num=3749
s.rom(num)
```

```python
class Solution(object):
    def intToRoman(self, num):
        dic={
            1000:"M",
            900:"CM",
            500:"D",
            400:"CD",
            100:"C",
            90:"XC",
            50:"L",
            40:"XL",
            10:"X",
            9:"IX",
            5:"V",
            4:"IV",
            1:"I",    
            }
        s=""
        for i in dic:
            while num>=i:
                s+=dic[i]
                num-=i
        return s
s=Solution()
num=3749
s.intToRoman(num)
```
