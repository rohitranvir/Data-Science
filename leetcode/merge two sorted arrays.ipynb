```python
class Solution(object):
    def mergeTwoLists(self, list1, list2):
        res=[]
        i=j=0
        while i<len(list1) and j<len(list2):
            if list1[i]<list2[j]:
                if not res or res[-1]!=list1[i]:
                    res.append(list1[i])
                i+=1
            elif list1[i]>list2[j]:
                if not res or res[-1] !=list2[j]:
                    res.append(list2[j])
                j+=1
            else:
                if not res or res[-1]!=list1[i]:
                    res.append(list1[i])
                    i+=1
                    j+=1
        while i<len(list1):
            if res[-1]!=list1[i]:
                res.append(list1[i])
                i+=1
        while j<len(list2):
            if res[-1]!=list2[j]:
                if list2[j] not in res:
                    res.append(list2[j])
                j+=1
        return res
s=Solution()
n1=[1,1,1,2,4,6,7]
n2=[1,2,3,6,7,8,9,10]
s.mergeTwoLists(n1, n2)
```
