```python
nums = [0,1,0,3,12]

n1=[]
zero_count=0
for i in nums:
    if i==0:
        zero_count+=1
    else:
        n1.append(i)
n1.extend([0]*zero_count)
print(n1)
```

```python
nums = [0,1,0,3,12]
k=0
for i in range(len(nums)):
    if nums[i]!=0:
        nums[k]=nums[i]
        k+=1
for j in range(k,len(nums)):
    nums[j]=0
print(nums)
```

```python
nums = [0,1,0,3,12]
for i in nums:
    if i==0:
        nums.remove(i)
        nums.append(i)
print(nums)
```

```python
nums = [0,1,0,3,12]
numsofzero=[]
while 0 in nums:
    nums.remove(0)
    numsofzero.append(0)
nums.extend(numsofzero)
print(nums)
```
